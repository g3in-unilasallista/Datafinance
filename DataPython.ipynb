# 2. Importación de Librerías
Importamos las librerías utilizando alias (como `yf` o `px`). Esta es una buena práctica en Python que hace que el código sea más **limpio, legible y profesional.**


````python
import plotly.express as px
import yfinance as yf
import google.generativeai as genai
import pandas as pd

## nbformat

print("Librerias importadas correctamente.")
````

## 3. Obtención de Datos Financieros

Definimos los *tickers* (códigos de bolsa) de las empresas y descargamos sus datos históricos del último año (`period="1y"`). Extraemos específicamente la columna de precios de cierre (`Close`).

````python
# 1. Crear los Tickers de las empresas
tickers = ["AAPL", "NVDA", "MSFT"]

# 2. Descargar los datos
datos = yf.download(tickers, period="1y")

# 3. Extraer solo los precios de cierre
precios = datos["Close"]

print("Datos descargados ok")
````

# 4. Exploración y validación de los datos
Como analista de datos, siempre debemos verificar que los datos se cargaron correctamente. Para ello, podemos utilizar las siguientes funciones y métodos:

- .head(): Permite visualizar la cabecera y los primeros registros del DataFrame.

- .tail(): Permite ver los últimos registros del conjunto de datos.

- .describe(): Permite obtener estadísticas descriptivas automáticas de las columnas numéricas.

**Nota:** Veremos aproximadamente una cantidad específica de filas. Debido a que no se van a mostrar todos los datos de los 365 días del año (sino solo los días de la semana hábiles), no se visualizan los días festivos acorde a esta exploración y validación de datos.

# 5. Gráfico Interactivo de Líneas (Cotización Histórica)
Creamos un gráfico de líneas profesional utilizando `plotly.express`. Añadimos un título claro, nombres personalizados para los ejes y un tema oscuro (`plotly_dark`) para una mejor visualización.

````python
# Crea un grafico de lineas interactivo
fig_linea = px.line(
    precios,
    title="Cotización Histórica: Apple, Microsoft & Nvidia (Último Año)",
    labels={"Date": "Fecha", "value": "Precio de Cierre U$"},
    template="plotly_dark"
    )

# Muestre el grafico
fig_linea.show()
````

# 6. Gráfico de Barras (Precio Promedio)
Calculamos la media de los precios de cada empresa y la visualizamos en un **gráfico de barras**, asignando un color diferente a cada empresa para mejorar la legibilidad.

````python
# Calcular la media de los precios y presentar un índice para que sea una columna graficable
precios_promedio = precios.mean().to_frame().reset_index()
precios_promedio.columns = ["Empresa", "Precio Promedio"]

# Crear el Gráfico de Barras
fig_barras = px.bar(
    precios_promedio,
    x="Empresa",
    y="Precio Promedio",
    color="Empresa",
    title="Precio Promedio: Apple, Microsoft & Nvidia (Último Año)",
    labels={"Precio Promedio": "Precio Promedio U$", "Empresa":"Empresa"},
    template="plotly_dark",
    text_auto='.2f'
)

# Mostrar el Gráfico
fig_barras.show()
````

# 7. Análisis de Inversión ($1,000 USD)

Simulamos qué habría pasado si un cliente hubiera invertido $1,000 USD en cada una de estas empresas al inicio del periodo. La fórmula es:

> **Inversión = (Precio actual / Precio inicial) * 1000**


````python
# Calculo del valor de la inversion en lo largo del tiempo
inversion = (precios / precios.iloc[0]) * 1000

# Crear el grafico de la inversión
fig_inversion = px.line(
    inversion,
    title="Comportamiento de la inversión que inicia en $1.000 USD",
    labels={"Date": "Fecha", "value": "Inversión en USD"},
    template="plotly_dark"
)

# Mostrar el gráfico
fig_inversion.show()
````

# 8. Exportar el Gráfico a `HTML`

Para que el cliente o el jefe pueda ver el gráfico interactivo sin necesidad de instalar Python, lo exportamos como un archivo `.html`. Este archivo se puede abrir en cualquier navegador web.

````python
# Guardar el grafico como HTML
fig_linea.write_html("cotizacion_historica.html")
fig_barras.write_html("precio_promedio.html")
fig_inversion.write_html("inversion.html")

print("Graficos exportados ok")
````

## 9. Informe Automatizado con Inteligencia Artificial (Google Gemini)

Utilizamos la API de Google Gemini para generar un informe ejecutivo en lenguaje natural. La IA analizará los datos y buscará justificar las alzas y bajas con base en eventos y noticias del mercado.

> **⚠️ IMPORTANTE:** Debes reemplazar `"TU_API_KEY_AQUI"` con tu propia clave de API de Google Gemini. Puedes obtenerla gratis en [Google AI Studio](https://aistudio.google.com/app/apikey).

````python
# Esta sección ahora simulará lo que la IA haría, sin llamar a la API.
# Ya no necesitamos configurar la API de Gemini ni inicializar el modelo.

# 1. Preparar los datos para el prompt (usamos un resumen o los últimos datos para no saturar el contexto)
# Estos datos son los que se habrían enviado a la IA para su análisis.
print("Preparando los datos que serían enviados a la IA...")
datos_resumen = precios.describe().to_string()
ultimos_datos = precios.tail(10).to_string()

# 2. Crear el prompt (la instrucción para la IA)
# Este es el texto que le diría a la IA qué tipo de informe generar y con qué información.
prompt = f"""
Actúa como un analista financiero experto.
Haz un análisis ejecutivo del último año de las acciones de Apple (AAPL), Microsoft (MSFT) y Nvidia (NVDA).

Estadísticas generales del periodo:
{datos_resumen}

Comportamiento reciente (últimos 10 días de cotización):
{ultimos_datos}

Tu informe debe incluir:
1. Un resumen general del rendimiento de cada empresa.
2. Posibles eventos, noticias o tendencias del mercado que justifiquen las alzas y bajas más relevantes observadas en los datos.
3. Una conclusión o recomendación breve para un inversor a largo plazo.

Usa un formato profesional, claro y con viñetas (Markdown).
"""

# 3. Mostrar el prompt que se habría enviado a la IA y explicar lo que se espera.
print("\n" + "="*50)
print("📄 SIMULACIÓN DE INFORME GENERADO POR IA")
print("="*50 + "\n")
print("Este es el prompt (la instrucción) que se habría enviado al modelo de IA para generar el informe:\n")
print("```markdown")
print(prompt)
print("```\n")
print("\nLa IA habría utilizado las 'Estadísticas generales del periodo' y el 'Comportamiento reciente' de los datos financieros proporcionados para:")
print("- Resumir el rendimiento de AAPL, MSFT y NVDA durante el último año.")
print("- Intentar identificar eventos o tendencias del mercado que pudieran explicar los movimientos de precios (esto requeriría conocimiento externo por parte de la IA).")
print("- Ofrecer una recomendación de inversión a largo plazo basada en su análisis.")
print("\nLa respuesta de la IA se habría formateado profesionalmente con viñetas en Markdown, tal como se especifica en el prompt.")
````
